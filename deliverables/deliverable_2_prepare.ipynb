# Phase 2: Prepare - Data Sources

## Data Description
For this analysis, I utilized Cyclistic’s historical trip data covering the last 12 months.

* **Primary Source:** Data provided by Motivate International Inc.
* **Data Organization:** The dataset is organized in monthly CSV files. Each file contains 13 columns, including ride ID, rideable type, start/end timestamps, station information, and member/casual classification.
* **Credibility (ROCCC):**
    * **Reliable:** High-volume dataset representing millions of real trips.
    * **Original:** Collected directly via the bike-sharing system.
    * **Comprehensive:** Covers all bike types and user segments.
    * **Current:** Analysis uses the most recent 12 months of historical data.
    * **Cited:** Provided under a public license by a reputable source.
* **Licensing and Privacy:** The data is public under a specific [Data License Agreement](https://www.divvybikes.com/data-license-agreement). No Personally Identifiable Information (PII) is included, ensuring user privacy.
