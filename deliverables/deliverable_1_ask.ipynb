# Phase 1: Ask - Business Task

## Problem Statement
The primary goal of this analysis is to identify how annual members and casual riders use Cyclistic bikes differently. These insights will be used to design a marketing strategy aimed at converting casual riders into annual members.

## Business Task
Identify key differences between casual riders and annual members to provide data-driven recommendations for a targeted marketing campaign that maximizes annual memberships.

## Stakeholders
* **Lily Moreno:** Director of Marketing, responsible for campaigns.
* **Cyclistic Marketing Analytics Team:** Responsible for data collection and analysis.
* **Cyclistic Executive Team:** The decision-makers who will approve the marketing program.
