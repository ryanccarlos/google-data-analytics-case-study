## About the Case Study: Cyclistic Bike-Share Analysis
In this capstone project for the Google Data Analytics Professional Certificate, I acted as a Junior Data Analyst for Cyclistic, a fictional bike-share company in Chicago. The goal was to analyze over 5.5 million trip records to identify behavioral differences between annual members and casual riders. 

By following the official data analysis process (Ask, Prepare, Process, Analyze, Share, and Act) and utilizing Python for large-scale data manipulation and visualization, I uncovered that members use the service for commuting while casuals use it for leisure. My final deliverables include data-driven recommendations to convert casual riders into loyal annual members through targeted seasonal campaigns and functional incentives.
