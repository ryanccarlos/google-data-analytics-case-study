# Phase 3: Process - Documentation of Cleaning and Manipulation

## Tools Used
* **Python (Pandas & NumPy):** Chosen for its ability to handle over 5.5 million rows efficiently and for reproducible data manipulation.

## Data Cleaning and Transformation Steps
1. **Consistency Check:** Verified that all monthly files had identical column headers and data types.
2. **Date Conversion:** Converted `started_at` and `ended_at` columns from strings to datetime objects.
3. **Metric Creation:**
    * **`ride_length`:** Calculated the duration of each trip in minutes (`ended_at` - `started_at`).
    * **`day_of_week`:** Extracted the day of the week from the start date (Mapped where 1 = Sunday and 7 = Saturday).
4. **Data Filtering:**
    * Removed rows where `ride_length` was less than or equal to 0 minutes (system errors/maintenance).
    * Removed outlier rows where `ride_length` exceeded 24 hours (1440 minutes).
5. **Validation:** Used `annual_df.info()` and `annual_df.describe()` to ensure data integrity and the absence of illogical values.
